#### Parametrized Constructor

- Constructor with parameter

```python
class Student:
    def __init__(self,name,age,sem):
        self.name=name
        self.age=age
        self.sem=sem

# self use hunxa objectko attributes define garna
s1=Student("Puja",21,"5th")
s2=Student("Deepa",22,"5th")


print(s1.name,s1.age,s1.sem)
print(s2.name,s2.age,s2.sem)


     
```

```python
### Memory Proof
print(id(s1))
print(id(s2))
```

#### Encapsulation

- Data abstraction
- Access Modifiers (public.protected,private)
- getter/setter


- __ makes a attribute private
- _ makes a attribute protected


### Why ?
- Prevent unauthorized access
- Controlled interaction

```python

## without encapsulation (private access modifier)
class BankAccount:

    def __init__(self,balance):
        self.balance = balance

acc = BankAccount(100000)

acc.balance = -50000

print(acc.balance)



class Account:
    def __init__(self,balance):
       self.__balance= balance ### private

    def get_balance(self):
        return self.__balance
    
acc = Account(5000)
print(acc.get_balance())
```

```python
## getter and setter both uses 

class BankAccount:
    def __init__(self,balance):
        self.__balance=balance

    def set_balance(self,amount):

        if amount >= 0:
            self.__balance= amount
        else:
            print("Invalid amount")

    def get_balance(self):
        return self.__balance
    
acc= BankAccount(0)

acc.set_balance(-500)
# acc.set_balance(500)
print(acc.get_balance())



    
```

#### Inheritance 

- Inheritance is the feature of OOPs where the child inherits the property of parent class

##### Why?

- Code reusability
- Code clean architecture
- easy maintenance
- Real world modeling



#### Single Inheritance

- Parent class -> child class

```python
## Parent class(base class)

class Employee:
    def login(self):
        print("Login successfully")

## single inheritance(Employee -> Developer)

class Developer(Employee):
    def code(self):
        print("Writes code")

dev = Developer()
dev.login()
dev.code()
```

#### Multilevel inheritance
- Grandparent -> parent -> child

```python
## Grandparents
class Employee:
    def login(self):
        print("Login successfully")


## Parents
class TechnicalEmployee(Employee):
    def access_server(self):
        print("Accessing server")

## child class
class Developer(TechnicalEmployee):
    def write_api(self):
        print("Writting API")

## creating object
dev = Developer()
dev.login()
dev.access_server()
dev.write_api()
```

#### Hierarchical Inheritance

- one parent -> multiple child

```python
## parent
class Employee:
    def login(self):
        print("Login successfully")

class Developer(Employee):
    def write_api(self):
        print("Writting API")

class Manager(Employee):
    def manage(self):
        print("Managing the team")


dev=Developer()
dev.login()
dev.write_api()


mgr= Manager()
mgr.login()
mgr.manage()

```
