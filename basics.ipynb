```python
# operators + how to take user input?

age = int(input("Enter the age "))

if age>0 and age<18:
    print(f"{age} is a child ")
elif age>18 and age<60:
    print(f"{age} is an adult")
elif age>60 and age<100:
    print(f"{age} is an old")
else:
    print(f"{age} is invalid age")
```

### Looping control statements

- It repetedly execute the given set of statement until the condition is satisfied
- Entry control loop : while
- For loop: intital bata kati samma janxa tha vako case ma use hunxa
- For each loop (list,array)

```python
# while loop
count =0
while(count <=10):
    count += 1
    print(f"{count} Hello Python Basics")
else:
    print("No Python")
```

```python
# for loop

n=5
for i in range(0,n):
    print(i, end=",")
```

```python
# *
# **
# ***
# ****
# *****
n=5
for i in range(1,n+1):
    print("*" * i)
```

```python
# *****
# ****
# ***
# **
# *
n=5
for i in range(n,0,-1):
    print("*" * i)
```

### List
- List is like an arrays (index), in other programming language
- It can support multiple datatypes
- List are mutable


```python
## List

fruits= ["apple","mango","banana","grapes","watermelon"]
print(fruits)

# list of mixed datatype
list =[1,"Puja","Deepa",89,10.23,11,45.6]
print(list)


# accessing list element
print(fruits[2])

# slicing
print(fruits[0:2])
print(fruits[-1:])


# negative indexing
print(fruits[-1])
print(fruits[-2])

```

### Tuples:
- Just like a list a tuple is a ordered collection of object in python.
- The difference is they are immutuable(We cannot modify it once created)

```python
# empty tuple 
a=()

t1=("Python", "Java")
print(t1)




t2 = (4,6,7)
t3 = t1 + t2
print(t3)

```

### Dictionary 
- Dictionary is a collection of different data types stored in value in key value pair
- Key and values are separated by : (colon) example: ("id": 101)
- , separated multiple values

```python
data = {
    1: "Python",
    2:" Programming",
    3: " Language"
}
print(data)
```
